## Flatten Binary Tree to Linked List

Given the root of a binary tree, flatten it into a linked list in-place. The flattened tree should use the same TreeNode class where the right child points to the next node in the list and the left child is always null. The order should follow preorder traversal `(root -> left -> right)`.

__Example 1:__

**Input:** 
`{"root":[1,2,5,3,4,null,6]}`

**Output:** 

`[1,null,2,null,3,null,4,null,5,null,6]

Example 2:
Input: 
{"root":[1,null,2,null,3]}
Output: 
[1,null,2,null,3]

Constraints:
The number of nodes in the tree is in the range [0, 2000]
-100 <= Node.val <= 100
