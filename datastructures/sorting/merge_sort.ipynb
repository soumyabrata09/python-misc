```python
from enum import Enum
from typing import List

class OrderBy(Enum):
    ASC = 'ascending'
    DESC = 'descending'

class MergeSort:

    @staticmethod
    def validate(arr: List[int]) -> None:
        if not arr:
            raise ValueError('Invalid array')

    @staticmethod
    def sort_by_asc(arr: List[int]) -> None:
        MergeSort.validate(arr)
        MergeSort.divide_and_conquer(arr, OrderBy.ASC)

    @staticmethod
    def sort_by_desc(arr: List[int]) -> None:
        MergeSort.validate(arr)
        MergeSort.divide_and_conquer(arr, OrderBy.DESC)

    @staticmethod
    def divide_and_conquer(arr: List[int], order: OrderBy) -> None:
        if len(arr) < 2: # Stopping condition for recursion
            return;

        mid: int = len(arr) // 2
        left: List[int] = arr[:mid] # creating left array from 0 to mid
        right: List[int] = arr[mid:] # creating right array from mid to len(arr)

        MergeSort.divide_and_conquer(left, order)
        MergeSort.divide_and_conquer(right, order)
        MergeSort.merge(arr, left, right, order)

    @staticmethod
    def merge(original_arr: List[int], left_sub_arr: List[int], right_sub_arr: List[int], order: OrderBy) -> None:
        arr_index: int = 0
        left_index: int = 0
        right_index: int = 0
 
        while left_index < len(left_sub_arr) and right_index < len(right_sub_arr):
            should_take_left: bool = (
                left_sub_arr[left_index] <= right_sub_arr[right_index]
                if order == OrderBy.ASC
                else left_sub_arr[left_index] >= right_sub_arr[right_index]
            )
            if should_take_left:
                original_arr[arr_index] = left_sub_arr[left_index]
                left_index += 1
            else:
                original_arr[arr_index] = right_sub_arr[right_index]
                right_index += 1
                
            arr_index += 1

        while left_index < len(left_sub_arr):
            original_arr[arr_index] = left_sub_arr[left_index]
            left_index += 1
            arr_index += 1

        while right_index < len(right_sub_arr):
            original_arr[arr_index] = right_sub_arr[right_index]
            right_index += 1
            arr_index += 1

arr = [13, 8, 51, 3, 5, 31, 9, 2]
arr_asc = arr.copy()
MergeSort.sort_by_asc(arr_asc)
print(f'sorted array in ascending order: {arr_asc}')

arr_desc = arr.copy()
MergeSort.sort_by_desc(arr_desc)
print(f'sorted array in descending order: {arr_desc}')
```
