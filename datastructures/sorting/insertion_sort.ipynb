Insertion Sort

## Idea

Insertion Sort builds the sorted array one element at a time.

Similar to arranging playing cards in your hand.

### Steps

1. Pick the next element (key).
2. Compare with elements on its left.
3. Shift larger elements one position right.
4. Insert the key into its correct position.

## Complexity

| Case | Time Complexity |
|--------|--------|
| Best (already sorted) | O(n) |
| Average | O(n²) |
| Worst (reverse sorted) | O(n²) |

### Space Complexity

```python
O(1)
```

````python
from typing import List

class InsertionSort:
    def __init__(self):
        pass

    def sort_by_asc(self, arr: List[int]) -> None:
        for inspected_index in range(1, len(arr)):
            inspected_curr_val = arr[inspected_index]
            prev_index = inspected_index - 1

            while prev_index >= 0 and arr[prev_index] > inspected_curr_val:
                arr[prev_index+1] = arr[prev_index]
                prev_index -= 1
            
            arr[prev_index+1] = inspected_curr_val

    def sort_by_desc(self, arr: List[int]) -> None:
        for inspected_index in range(1, len(arr)):
            inspected_curr_val = arr[inspected_index]
            prev_index = inspected_index - 1

            while prev_index >= 0 and arr[prev_index] < inspected_curr_val:
                arr[prev_index+1] = arr[prev_index]
                prev_index -= 1
            
            arr[prev_index+1] = inspected_curr_val

arr = [13, 8, 51, 3, 5, 31, 9, 2]

arr_asc = arr.copy()
arr_desc = arr.copy()

obj = InsertionSort()

obj.sort_by_asc(arr_asc)
print(f'Sorted array[Asc]: {arr_asc}\n')
obj.sort_by_desc(arr_desc)
print(f'\nSorted array[desc]: {arr_desc}')
````
