### Why?

#### 🟢 Best Case — `O(n log n)`

* Pivot divides the array into two nearly equal halves every time.
* Recursion depth is approximately `log₂(n)`.
* Each level processes all `n` elements.

```text
Level 0 : n
Level 1 : n
Level 2 : n
...
log₂(n) levels

Total = n × log n
```

---

#### 🟢 Average Case — `O(n log n)`

For random input, partitions are reasonably balanced most of the time.

```text
Random Input
     ↓
Balanced partitions
     ↓
O(n log n)
```

---

#### 🔴 Worst Case — `O(n²)`

Occurs when the pivot is always the smallest or largest element.

Since your implementation always chooses:

```python
pivot = arr[high]
```

The following inputs trigger the worst case:

Ascending sorted input

```text
1 2 3 4 5 6 7
```

Descending sorted input

```text
7 6 5 4 3 2 1
```

Each recursive call reduces the problem by only one element:

```text
n
↓
n-1
↓
n-2
↓
...
↓
1
```

Total work:

```text
n + (n-1) + (n-2) + ...

= n(n-1)/2

= O(n²)
```

---

## 📦 Space Complexity

Quick Sort is **in-place**, meaning it does **not** allocate additional arrays during partitioning.

However, recursive calls consume stack space.

### Best / Average

Balanced recursion:

```text
          n
       /     \
     n/2     n/2
    /  \     /  \
```

Depth:

```text
log₂(n)
```

So,

> 🟢 **Space = O(log n)**

---

### Worst Case

Unbalanced recursion:

```text
n
|
n-1
|
n-2
|
...
|
1
```

Depth:

```text
n
```

So,

> 🔴 **Space = O(n)**

## 💡 Interview Tip

When asked about Quick Sort in interviews, a complete answer is:

> **Quick Sort is an in-place, divide-and-conquer sorting algorithm. It has an average and best-case time complexity of `O(n log n)` but degrades to `O(n²)` when partitions become highly unbalanced (for example, when always choosing the last element as the pivot on an already sorted array). Its auxiliary space complexity is `O(log n)` on average due to recursion, but it can grow to `O(n)` in the worst case. It is not a stable sorting algorithm.`**

This is the level of explanation typically expected in senior Java developer interviews.


#### In Quick Sort there are 4 ways to choose the pivot from the given array

- *Select any random data from the Array*
- *Select the middle value from the Array -> `low + high // 2`*
- *Select the 1st element of the Array*
- *Select the last element of the Array*

*Here in our solution we are going to select the last elemnt of the Array as our pivot*

````python
from enum import Enum
class OrderBy(Enum):
    ASC = 'ascending'
    DESC = 'descending'    

class QuickSort:

    @staticmethod
    def validate(arr: list[int]) -> None:
        if arr is None:
            raise ValueError('Not an Array or may be it is null')
        
    @staticmethod
    def _swap(arr: list[int], source_pos: int, target_pos: int) -> None:
        if source_pos == target_pos:
            return
        
        arr[target_pos], arr[source_pos] = arr[source_pos], arr[target_pos]

    @classmethod
    def sort_by_asc(cls, arr: list[int]) -> None:
        QuickSort.validate(arr)
        cls._single_pivot_quick_sort(arr, 0, len(arr) - 1, OrderBy.ASC)

    @classmethod
    def sort_by_desc(cls, arr: list[int]) -> None:
        QuickSort.validate(arr)
        cls._single_pivot_quick_sort(arr, 0, len(arr) - 1, OrderBy.DESC)

    @classmethod
    def _single_pivot_quick_sort(cls, arr: list[int], low: int, high: int, order: OrderBy) -> None:
        if low < high:
            pivot_index = cls._partition(arr, low, high, order)
            cls._single_pivot_quick_sort(arr, low, pivot_index - 1, order)
            cls._single_pivot_quick_sort(arr, pivot_index + 1, high, order)
    
    @classmethod
    def _partition(cls, arr: list[int], low: int, high: int, order: OrderBy) -> int:
        partition_index: int = low
        pivot: int = arr[high]

        for current_index in range(low, high):
            if order == OrderBy.ASC:
                if arr[current_index] < pivot:
                    # swap and increment `partition_index`
                    QuickSort._swap(arr, partition_index, current_index)
                    partition_index += 1
            else:
                if arr[current_index] > pivot:
                    # swap the elem and increment the `spartition_indextart`
                    QuickSort._swap(arr, partition_index, current_index)
                    partition_index += 1
        
        # swap `partition_index` and high
        QuickSort._swap(arr, partition_index, high)

        return partition_index


arr = [13, 8, 51, 3, 3, 5, 31, 9, 2]
arr_asc = arr.copy()
arr_desc = arr.copy()

QuickSort.sort_by_asc(arr_asc)
print(f'Sorted array in ascending order -> {arr_asc}')

QuickSort.sort_by_desc(arr_desc)
print(f'Sorted array in descending order -> {arr_desc}')
````
