## 1. Selection Sort

### Idea

Selection Sort repeatedly finds the smallest element from the unsorted portion and places it at its correct position.

### Steps

1. Find the minimum element in the unsorted part.
2. Swap it with the first unsorted position.
3. Move the boundary of the sorted portion one step right.
4. Repeat until the array is sorted.

---

### Example

Initial Array:

```python
[64, 25, 12, 22, 11]
```

Legend:

- ✅ Sorted portion
- ⭐ Current minimum

````python
from typing import List

class SelectionSort:
    def __init__(self):
        pass

    def sort_by_asc(self, arr: List[int]) -> None:
        for inspected_index in range(len(arr)-1):
            min_index = inspected_index
            for j in range(inspected_index+1, len(arr)):
                if arr[min_index] > arr[j]:
                    min_index = j
                
            self.swap(arr, inspected_index, min_index)

    def sort_by_desc(self, arr: List[int]) -> None:
        for inspected_index in range(len(arr)-1):
            max_index = inspected_index
            for j in range(inspected_index+1, len(arr)):
                if arr[max_index] < arr[j]:
                    max_index = j

            self.swap(arr, inspected_index, max_index)
    
    def swap(self, arr: List[int], target_pos: int, source_pos: int) -> None:
        arr[target_pos], arr[source_pos] = arr[source_pos], arr[target_pos]

arr = [13, 8, 51, 3, 5, 31, 9, 2]

arr_asc = arr.copy()
arr_desc = arr.copy()

obj = SelectionSort()

obj.sort_by_asc(arr_asc)
print(f'Sorted array[Asc]: {arr_asc}\n')
obj.sort_by_desc(arr_desc)
print(f'\nSorted array[desc]: {arr_desc}')
````
