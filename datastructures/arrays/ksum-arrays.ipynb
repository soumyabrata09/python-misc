18. 4Sum [Leetcode]

> Given an array nums of n integers, return an array of all the unique quadruplets [nums[a], nums[b], nums[c], nums[d]] such that:

`0 <= a, b, c, d < n`<br />
a, b, c, and d are distinct.<br />
`nums[a] + nums[b] + nums[c] + nums[d] == target`<br />
You may return the answer in any order.<br />

__Example 1:__

Input: `nums = [1,0,-1,0,-2,2]`, `target = 0` <br />
Output: [[-2,-1,1,2],[-2,0,0,2],[-1,0,0,1]] <br />

__Example 2:__

_Input:_ `nums = [2,2,2,2,2]`, `target = 8`<br />
_Output:_ `[[2,2,2,2]]`
 

<div style="padding:12px;border-left:6px solid #FF9800;background:#FFF3E0;">
<b>⚠️ Warning</b><br>

> Constraints

> 1 <= nums.length <= 200<br />
> -109 <= nums[i] <= 109<br />
> -109 <= target <= 109<br />
</div>

```python
class KSumArrays:

    @staticmethod
    def _k_sums(nums: list[int], target: int, k: int, start: int) -> list[list[int]]:
        if start >= len(nums):
            return []
        
        # Pruning clause
        min_possible = nums[start] * k
        max_possible = nums[len(nums) - 1] * k

        if target < min_possible or target > max_possible:
            return []
        
        result: list[list[int]] = []

        if k == 2: # Two-Sum Approach
            left = start
            right = len(nums) - 1

            while left < right:
                current_sum = nums[left] + nums[right]
                if current_sum == target: # Match found
                    result.append([
                        nums[left], nums[right]
                    ])

                    while left < right and nums[left] == nums[left + 1]:
                        left += 1
                    while left < right and nums[right] == nums[right - 1]:
                        right -= 1

                    left += 1
                    right -= 1
                elif target > current_sum:
                    left += 1
                else:
                    right -= 1
        else:
            for current_index in range(start, len(nums)):
                temp_result = KSumArrays._k_sums(nums, target - nums[current_index], k - 1, current_index + 1)

                if temp_result:
                    for temp in temp_result:
                        temp.insert(0, nums[current_index])
                    
                    result.extend(temp_result)
                
                while current_index < len(nums) - 1 and nums[current_index] == nums[current_index + 1]:
                    current_index += 1
        
        return result

    @staticmethod
    def four_sums(nums: list[int], target: int = 4) -> list[list[int]]:
        if not nums or len(nums) < 4:
            return []
        nums.sort()
        return KSumArrays._k_sums(nums, target, 4, 0)
    
nums = [1, 0, -1, 0, -2, 2]
resultant_arr = KSumArrays.four_sums(nums, 0)
print(f'Resultant Array of {nums} is as follows\n\t{resultant_arr}')
```
