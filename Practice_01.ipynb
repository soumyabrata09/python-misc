## Python class Demo

````python
class TextAnalyzer(object):
    
    def __init__ (self, text):
        self.text = text;
        # remove punctuation
        formatted_txt = self.text.replace('!', '').replace(',', '').replace('.', '').replace('?', '').lower()
        # make text lowercase
        self.frmt_text = formatted_txt;
    def freq_all(self):        
        # split text into words
        words = self.frmt_text.split();
        # Create a dictionary
        word_freq = {};
        for word in words:
            word_freq[word] = words.count(word);
        return word_freq;
           
    def freq_of(self, word):
        # get frequency map
        word_freq = self.freq_all();
        if (word in word_freq):
            return word_freq[word];
        else: 
            return 0; 
````

````python
given_string="Lorem ipsum dolor! diam amet, consetetur Lorem magna. sed diam nonumy eirmod tempor. diam et labore? et diam magna. et diam amet.";
textAnalyzer = TextAnalyzer(given_string);
print("Formatted Text: ", textAnalyzer.frmt_text);
print(textAnalyzer.freq_all());
print(textAnalyzer.freq_of("lorem"));
````

### File write operation demo

**File copy utility**

````python
class FileCopyUtil:
    def __init__(self, source_file):
        self.source_file = source_file
        self.name = None
        self.mode = None

    def copy(self, dest_file):
        try:
            with open(self.source_file, "r") as sourceFile:
                with open(dest_file, "w") as destFile:
                    for line in sourceFile:
                        destFile.write(line)

                    self.name = destFile.name
                    self.mode = destFile.mode

            print("File has been copied successfully")
        except IOError as e:
            print("Unable to copy the file", e)
             
    
    def show_metadata(self):
        if (self.name and self.mode):
            print(f"File Name: {self.name}, mode: {self.mode}")

from pathlib import Path
import math

file_copy_util = FileCopyUtil("./TextAnalyzer.py")
destination_file = file_copy_util.source_file + '_copy.py'
file_copy_util.copy(destination_file)
file_copy_util.show_metadata()
dest_file_path = Path(destination_file)
file_size_bytes = dest_file_path.stat().st_size
file_size = file_size_bytes / (math.pow(1024, 2))
print("Source File exists ? ", Path(file_copy_util.source_file).exists())
print(f"File Size: {file_size:.2f} MB")
````

## Lambda Demo
<div class="alert alert-block alert-success">
    Lambdas are useful for small one liner small block of codes.
</div>

***Syntax***
```python
variable_name = lambda expression
```
> Usage: filter(), reduce(), sorted(), map() etc.

````python
"""
Example usage of reduce() method to get the product of the example List
"""
from functools import reduce

lst = [3, 4.15, 3.9, 7, 2.3, 6, 12]
product = reduce(lambda x, y: (x * y), lst)
print(product)
````

````python
"""
Lambda with if-else conditional branching
"""
text = 'Lorem Ipsum, dolores!'
checkIsSubSetOf = lambda word: True if set(word.lower()).issubset(text) else False

print(checkIsSubSetOf('rem'))
print(checkIsSubSetOf('xyz'))
````

````python
"""
Example usage of map() method
"""
import math

print(lst)
# suare root of each element of the list
sqrt_lst = list(map(lambda x: math.sqrt(x), lst))
print(sqrt_lst)
````

````python
"""
Example usage of the filter() method
"""
print("Even: ", list(filter(lambda item: item % 2 == 0, lst)))
print("Odd: ", list(filter(lambda item: item % 2 == 1, lst)))
# print("Integers: ", list(filter(lambda item: type(item) == int, lst)))
print("Integers: ", list(filter(lambda item: isinstance(item, int), lst)))
# print("Floats: ", list(filter(lambda item: type(item) == float, lst)))
print("Floats: ", list(filter(lambda item: isinstance(item, float), lst)))

checkIntOrFloat = lambda value, data_type: isinstance(value, data_type)

# print("Sum of only floats: ", sum(filter(lambda item: isinstance(item, float), lst)))
print("Sum of only floats: ", sum(filter(lambda item: checkIntOrFloat(item, float), lst)))
print("Sum of only ints: ", sum(filter(lambda item: checkIntOrFloat(item, int), lst)))
print("No. of floats in the list: ",len((list(filter(lambda itm: checkIntOrFloat(itm, float), lst)))))
print("No. of ints in the list: ",len((list(filter(lambda itm: checkIntOrFloat(itm, int), lst)))))
````

````python
"""
Example of sorted() method using lambda
"""
city_temps = {
    'Kolkata': { 'min_temp': 21, 'max_temp': 29 },
    'Mumbai': { 'min_temp': 24, 'max_temp': 32 },
    'Delhi': { 'min_temp': 22, 'max_temp': 33 },
    'Bangalore': { 'min_temp': 19, 'max_temp': 27 },
    'Chennai': { 'min_temp': 27, 'max_temp': 35 }
}

# sort by max_temp
min_temp_sorted = sorted(city_temps.items(), key= lambda city: city[1]['min_temp'])
max_temp_sorted = sorted(city_temps.items(), key = lambda city: city[1]['max_temp'], reverse = True) # Highest max_temp will come first...

def print_city_temp(sorted_city_temps):
    for city, temp in sorted_city_temps:
        print(f"{city}: {temp}")

# print("Sorted by minimun temp. -> \n", dict(min_temp_sorted))
print("Sorted by minimun temp.")
print_city_temp(min_temp_sorted)

# print("Sorted by maximum temp. -> \n", dict(max_temp_sorted))
print("\nSorted by maximum temp. -> ")
print_city_temp(max_temp_sorted)
````

### Temerature Analyzer

````python
from enum import Enum

"""
Enumeration: python
"""
class TemperatureType(Enum):
    MIN_TEMPERATURE = 'min_temp'
    MAX_TEMPERATURE = 'max_temp'
````

````python
"""
Data
"""
city_temps = {
    'Kolkata': { 
        TemperatureType.MIN_TEMPERATURE.value: '23°c', 
        TemperatureType.MAX_TEMPERATURE.value: '29°c'
    },
    'Delhi': {
        TemperatureType.MIN_TEMPERATURE.value: '24°c',
        TemperatureType.MAX_TEMPERATURE.value: '31°c'
    },
    'Mumbai': {
        TemperatureType.MIN_TEMPERATURE.value: '25°c',
        TemperatureType.MAX_TEMPERATURE.value: '31°c'
    },
    'Bangalore': {
        TemperatureType.MIN_TEMPERATURE.value: '19°c',
        TemperatureType.MAX_TEMPERATURE.value: '28°c',
    },
    'Chennai': {
        TemperatureType.MIN_TEMPERATURE.value: '21°c',
        TemperatureType.MAX_TEMPERATURE.value: '35°c'
    }
}
````

<div class="alert alert-block alert-info">
    In Python too, we have access specifiers but it is very restrictive in nature like Java/C++. <br />
    <!-- <pre class="alert alert-block alert-warning"> -->

````python
class TemperatureAnalyzer:

    def __init__(self, city_temps):
        self.city_temps = city_temps
        self.sorted_min_temps = None
        self.sorted_max_temps = None

    """
    private sort method
    """
    def __sort(self, temp_type, desc_ordering = False):
        return sorted(city_temps.items(), key = lambda item: item[1][temp_type], reverse = desc_ordering)
    
    def sort_by_min(self):
        self.sorted_min_temps = self.__sort(TemperatureType.MIN_TEMPERATURE.value)
        return self.sorted_min_temps
    
    def sort_by_max(self):
        self.sorted_max_temps = self.__sort(TemperatureType.MAX_TEMPERATURE.value, True)
        return self.sorted_max_temps
    
    def get_min_temp(self):
        if (self.sorted_min_temps == None):
            return
        return self.sorted_min_temps[0]
    
    def get_max_temp(self):
        if (self.sorted_max_temps == None):
            return
        return self.sorted_max_temps[0]
    
    def show_temp(self, temp_type, sorted_temp):
        print(f"Temperature by {temp_type}\n")
        for temperature in dict(sorted_temp).items():
            print(f"{temperature[0]} => {temperature[1]}")
        print("=====\n")
````

````python
"""
Create object of TemperatureAnalyzer
"""

tempAnalyzer = TemperatureAnalyzer(city_temps)

tempAnalyzer.show_temp(TemperatureType.MIN_TEMPERATURE.value, tempAnalyzer.sort_by_min())
tempAnalyzer.show_temp(TemperatureType.MAX_TEMPERATURE.value, tempAnalyzer.sort_by_max())

print(f"Lowest Temperature metadata\n: {tempAnalyzer.get_min_temp()}")
print(f"\nHighest Temperature metadata\n: {tempAnalyzer.get_max_temp()}")
````
