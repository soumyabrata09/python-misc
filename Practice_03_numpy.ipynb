## What is NumPy? <hr />

`NumPy` is a Python library used for working with arrays, linear algebra, fourier transform, and matrices. `NumPy` stands for `Numerical Python` and it is an open source project. The array object in `NumPy` is called `ndarray`, it provides a lot of supporting functions that make working with ndarray very easy.

Arrays are very frequently used in data science, where speed and resources are very important.

NumPy is usually imported under the np alias.
---
It's usually fixed in size and each element is of the same type. We can cast a list to a numpy array by first importing numpy:

````python
import numpy as np
import random
import string

print(f"numpy version: {np.__version__}")

def random_number_generator(dtype = np.float64, length=5):
    if np.issubdtype(dtype, np.floating):
        return round(random.uniform(1, 10), 2)
    elif np.issubdtype(dtype, np.integer):
        return round(random.randint(1, 10))
    elif dtype == str:
        return ''.join(random.choices(string.ascii_letters, k=length))
    else:
        raise ValueError("Unsupported dtype")

# Create a numoy array
arr = np.array([2, -1, 4, 1, -5, 7])
print(f"ndarray: {arr}")
print(f"Size of ndarray: {arr.size}")

# Print all the elem of ndarray
index = 0
for index in range(arr.size):
    print(arr[index])
print(f"Type: {type(arr)}")
print(f"Data type: {arr.dtype}")

# Assign float value to {arr}
index = 0
arr = arr.astype(float) # Changed the dtype of ndarray from int64 to float64
# for index in range(arr.size):
#     arr[index] = random_number_generator()
#     print(f"Assigned value to ndarray: {arr[index]}")

# More pythonic way to assign the values 
arr = np.array(
    [random_number_generator() for _ in range(arr.size)]
)
print(f"After float value insertion, value of arr: {arr} and data-type: {arr.dtype}")
print(arr.ndim)
print(arr.shape)
print(arr.mean())
print(arr.min())
print(arr.max())
````

## Slicing <hr />
Like lists, we can slice the numpy array. Slicing in python means taking the elements from the given index to another given index.

We pass slice like this: `[start:end]`.The element at end index is not being included in the output.

We can select the elements from 1 to 3 and assign it to a new numpy array d as follows:

````python
print(f"Previously ndarray: {arr.view()}")
d = arr[1:4]
print(f"After slicing: {d.view()}")
````

__We can assign values though slicing__

````python
val1 = random_number_generator()
val2 = random_number_generator()
print(f"val1 = {val1} and val2 = {val2}")
# arr[2: 4]= val1, val2 # This is called broadcasting
# print(arr)
print(f"Previously d: {d.view()}")
d[1:3] = val1, val2
print(f"After broadcasting the values: {d.view()}")

# Broadcast to all the values of d
d = np.array([random_number_generator(np.integer) for _ in range(d.size)])
print(f"After broadcasting to all the elems of d[int]: {d.view()}")
d = np.array([random_number_generator(str, 6) for _ in range(d.size)])
print(f"After broadcasting to all the elems of d[str]: {d.view()}")
````

### More slicing techniques

 **Syntax**: `arr[start:end:step]`
 - __arr[:value]__ -> <span class="text-success">indicates that the slicing would start from the index 0 to give value(index) position</span>
 - __arr[value:]__ -> <span class="text-success">indicates that the slicing would start from the given index position till the end of the ndarray</span>
 - __arr[value_start:value_end:]__ -> <span class="text-success">indicates that the slicing would start from `value_start` till `value_end` with default stpe value as `1` if we do not specify the step value explicitly</span>


````python
arr_2 = np.array([random_number_generator(np.integer) for _ in range(10)])
print(f"arr_2: {arr_2.view()}")

print(arr_2[2:-1:2])
print(arr_2[:5]) 
print(arr_2[2:])
print(arr_2[1:-1:])
````

### Explanation of dot operation

<div class="alert alert-success">
<pre>
Let's say x = [2, 3] and y = [4, 2]
Hence,  x[0] = 2 and x[1] = 3,
        y[0] = 4 and y[1] = 2
<br />
Dot product will be -> 
x.y = [(x[0] * y[0]) + (x[1] + y[1])]
    = [(2*4) + (3*2)]
    = [8, 6]
</pre>
</div>

<div class="badge badge-warning">
The above operation would be as follows in the ndarray: 
</div><br /><br />

```python
import numpy as np

x = np.array([2, 3])
y = np.array([4, 2])

```

## Linspace <hr />

A useful function for plotting mathematical functions is <code>linspace</code>.   Linspace returns evenly spaced numbers over a specified interval. 

>```python
> import numpy as np
> np.linspace(start, stop, num = int value)
>```
 
- <span class="badge badge-success">__start__</span>  :  <span class="text-success">start of interval range

- <span class="badge badge-danger">__stop__</span>   :  <span class="text-danger">end of interval range</span>

- <span class="badge badge-warning text-dark">__num__</span>    :  <span class="text-warning">Number of samples to generate.</span>

````python
import matplotlib.pyplot as plt
import numpy as np
import string
# import pandas as pd
from enum import Enum

"""
Enum for available plot types
"""
class Plots(Enum):
    LINE_PLOT = 'line-plot'
    SCATTER_PLOT = 'scatter-plot'
    BAR_CHART = 'bar-chart'
    HISTOGRAM = 'histogram'
    PIE_CHART = 'pie-chart'

# Makeup a numpy array within [-2, 2] and 5 elements
print("With 5 samples, ", np.linspace(-2, 2, num=5))
print("With 9 samples, ", np.linspace(-2, 2, num=9))

"""
This methods rounds off the ndarray to a give decimal points
"""
def round_of(ndarray, decimal = 2):
    return np.round(ndarray, decimal)

"""
This method will add the caption to the plot grapgh
"""
def add_fig_caption(caption, positioning='center'):
    plt.figtext(0.50, -0.025, caption, ha = positioning)

"""
This method will add the plot characteristics values
"""
def add_plt_props(title, xlabel, ylabel, caption, position='center', show_legend=True):
    plt.title(title)
    plt.xlabel(f'x-plot {xlabel.view()}')
    plt.ylabel(f'y-plot {ylabel.view()}')
    plt.grid()
    add_fig_caption(caption, position)
    if show_legend:
        plt.legend()

# Make a numpy array within [0, 2π] and 15 elements 
# x = np.round(np.linspace(0, 2*np.pi, num=15), 2)
x = round_of(np.linspace(0, 2*np.pi, num=10), 2)
# print("x: ", x)
# y = np.round(np.sin(x), 2)
y = round_of(np.sin(x))
# print("y: ", y)

"""
Line Plot
"""
plt.plot(x, y, label="sin-plotting", color="brown", marker="o")
# plt.title("sin graph")
# plt.figtext(0.5, -0.025, "line-plot", ha="center")
# add_fig_caption(Plots.LINE_PLOT.value)
# plt.xlabel(f'x-plot {x.view()}')
# plt.ylabel(f'y-plot {y.view()}')
# plt.legend()
# plt.grid()
add_plt_props('sin-graph', x, y, Plots.LINE_PLOT.value) # Above commented lines moved inside the add_plt_props method
plt.show()

"""
Scatter Plot
"""
plt.scatter(x, y,label="sin-plotting", color="green", marker="x")
# plt.title("Sin graph")
# plt.figtext(0.5, -0.025, "scatter-plot", ha="center")
# add_fig_caption(Plots.SCATTER_PLOT.value)
# plt.xlabel(f'x-plot {x.view()}')
# plt.ylabel(f'y-plot {y.view()}')
# plt.legend()
# plt.grid()
add_plt_props('sin-graph', x, y, Plots.SCATTER_PLOT.value)
plt.show()

"""
Bar Chart
"""
plt.bar(x, y, color="purple")
# plt.title("Sin bar chart")
# plt.grid()
# plt.figtext(0.5, -0.025, "bar-chart", ha="center")
# add_fig_caption(Plots.BAR_CHART.value)
add_plt_props('Sin bar chart', x, y, Plots.BAR_CHART.value, 'center', False)
plt.show()

"""
Histogram
"""
# data = np.random.randn(100) # older way
# data = np.random.default_rng().standard_normal(100) # newer way (using Generator API)
data = x, y
print(data)
plt.hist(data, bins = 10, color=['orange', 'olive'])
# plt.grid()
add_plt_props('Sin histogram', x, y, Plots.HISTOGRAM.value)
plt.show()

"""
Pie Chart
"""
s = np.array(list(string.ascii_lowercase[:x.size])) # this will generate an ndarray as ['a', 'b', 'c', ...]
# s = pd.Series(list(string.ascii_lowercase[:x.size])).to_numpy() # using pandas
plt.pie(x, labels = s)
add_plt_props('Sin pie chart', x, y, Plots.PIE_CHART.value)
plt.show()

# y = np.cos(x)
# plt.plot(x, y)
# plt.show()

````
